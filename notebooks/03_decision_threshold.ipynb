# 03 — Business-aware threshold
Turn probabilities into decisions through an explicit cost model.

```python
thresholds = [
 {'threshold':.30,'false_negative':8,'false_positive':42},
 {'threshold':.50,'false_negative':18,'false_positive':21},
 {'threshold':.70,'false_negative':31,'false_positive':9}]
FN_COST, FP_COST = 10, 1
min(thresholds, key=lambda x: x['false_negative']*FN_COST + x['false_positive']*FP_COST)
```

Costs are scenario assumptions, not universal facts. Validate them with accountable stakeholders and keep human review for consequential decisions.
